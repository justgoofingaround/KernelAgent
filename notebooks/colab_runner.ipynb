```python
REPO = "https://github.com/justgoofingaround/KernelAgent.git"

import os, subprocess
try:
    from google.colab import userdata
    token = userdata.get("GITHUB_TOKEN")
    REPO = REPO.replace("https://", f"https://{token}@")
except Exception:
    pass  # public repo, no token needed

if os.path.isdir("/content/KernelAgent"):
    subprocess.run(["git", "-C", "/content/KernelAgent", "pull"], check=True)
else:
    subprocess.run(["git", "clone", REPO, "/content/KernelAgent"], check=True)
%cd /content/KernelAgent
```
